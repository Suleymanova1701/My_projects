## Описание проекта (#description_project)
<a id="description_project"></a>

Анализ популярного сервиса аренды самокатов `GoFast`. Задача — проанализировать данные и проверить некоторые гипотезы, которые могут помочь бизнесу вырасти.

Чтобы совершать поездки по городу, пользователи сервиса GoFast пользуются мобильным приложением. Сервисом можно пользоваться:
- **без подписки:**
    - абонентская плата отсутствует;
    - стоимость одной минуты поездки — 8 рублей;
    - стоимость старта (начала поездки) — 50 рублей;
- **с подпиской Ultra:**
    - абонентская плата — 199 рублей в месяц;
    - стоимость одной минуты поездки — 6 рублей;
    - стоимость старта — бесплатно.

*Цель исследования:*
1. Проанализировать данные о пользователях, подписках и поездках из представленных датасетов.
2. Найти `общие черты и особенности` представленных данных о частоте встречаемых городов, соотношении пользователей с подпиской и без, возраст пользователей, расстояние и продолжительность поездок.
3. Определить доли пользователей с подпиской и без.
4. Визуализировать информацию о расстоянии и времени поездок для пользователей обеих категорий.
5. Агрегация данных о поездках (суммарному расстоянию, количеству поездок и времени для каждого пользователя помесячно) с разбивкой по категориям пользователей.
6. Проверить гипотезы: по увеличению количества пользователей с подпиской, по оптимальному расстоянию одной поездки, сумме выручки пользователей с подпиской и без неё, снижению количества обращений в техподдержку после обновления серверов.
7. Проверить гипотезу по продлению подписки после проведённой акции, выяснить какое минимальное количество промокодов нужно разослать, чтобы вероятность не выполнить план была примерно 5%.
8. Построить примерный график распределения и оценить вероятность того, что уведомление в приложении от отдела маркетинга откроют не более 399,5 тыс. пользователей.

#### Функции для предобработки данных

```python
def get_count_and_share(df: pd.DataFrame, groupby: str = "city") -> pd.DataFrame:
    """
    Функция формирует и выводит сводную таблицу по одному из атрибутов (город по умолчанию),
    в которой отражает количество и долю строк каждого из этих атрибутов.
    """
    # считаю количество пользователей
    count_and_share = df.groupby(groupby).agg({groupby: "count"})

    # переименовываю столбец
    count_and_share.columns = ["count"]

    # считаю долю
    count_and_share["%_share_of_"] = round(
        (count_and_share["count"]).div(df["user_id"].count(), axis=0), 4
    )

    count_and_share = count_and_share[count_and_share["%_share_of_"] >= 0.001]

    # переименовываю столбцы финального датафрейма
    count_and_share.columns = ["count_of_" + groupby, "%_share_of_" + groupby]

    # сортирую финальный датафрейм по убыванию
    count_and_share = count_and_share.sort_values(
        by="%_share_of_" + groupby, ascending=False
    )

    count_and_share = count_and_share.style.format(
        {"%_share_of_" + groupby: "{:.2%}".format}
    ).background_gradient(cmap="Greens")
    return count_and_share
```

#### Функции для анализа и визуализации данных.

```python
# Функция display_pivot_table формирует сводные таблицы с неограниченным количеством параметров группировки
def get_pivot_table(
    data: pd.DataFrame, base_column: str, index_name: str
) -> pd.DataFrame:
    """
    Функция формирует и выводит сводную таблицу по одному из атрибутов,
    в которой отражает сумму и долю каждого из атрибутов.
    """

    pivot_table = data.pivot_table(
        index=index_name, values=base_column, aggfunc=["count"]
    ).reset_index()
    # переименовываю столбец
    pivot_table.columns = [index_name, "count_of_users"]

    # считаю долю платящих пользователей
    pivot_table["%_share_of_" + index_name] = (
        pivot_table["count_of_users"] * 100 / pivot_table["count_of_users"].sum()
    ).round(2)

    # сортирую финальный датафрейм по убыванию
    pivot_table = pivot_table.sort_values(by="count_of_users")

    return pivot_table
```

```python
# Функция для создания сводной таблицы с данными о поездках
def get_pivot_users_info(
    df: pd.DataFrame,
    index=["user_id", "month"],
    values=[
        "distance",
        "duration",
        "date",
        "minute_price",
        "start_ride_price",
        "subscription_fee",
        "subscription_type",
    ],
    aggfunc={
        "distance": "sum",
        "duration": "sum",
        "date": "count",
        "minute_price": "first",
        "start_ride_price": "first",
        "start_ride_price": "first",
        "subscription_fee": "first",
        "subscription_type": "first",
    },
) -> pd.DataFrame:
    pivot_table = (
        df.pivot_table(index=index, values=values, aggfunc=aggfunc)
        .reset_index()
        .rename(columns={"date": "rides_count"})
    )

    return pivot_table
```

```python
# Функция добавляет столбец с данными об общей выручке за каждый месяц поездок пользователей
def get_month_revenue(
    df: pd.DataFrame,
    start_ride_price: str,
    subscription_fee: str,
    duration: str,
    minute_price: str,
) -> pd.DataFrame:
    """
    Функция рассчитывает выручку помесячно
    """
    start_ride_price = df["start_ride_price"] * df["rides_count"]
    subscription_fee = df["subscription_fee"]
    duration = df["duration"]
    minute_price = df["minute_price"]

    df["revenue"] = start_ride_price + subscription_fee + duration * minute_price
    return df
```

```python
# Функция build_hist для генерации гистограмм по указанным параметрам
def build_hist(
    data: pd.Series,
    xname: str,
    title=None,
    xlabelsize=None,
    xrot=None,
    ylabelsize=None,
    yrot=None,
    ax=None,
    sharex=False,
    figsize=(20, 8),
    bins=40,
    legend=True,
):
    data.hist(figsize=figsize, bins=bins, legend=legend)
    plt.xlabel(xname)
    plt.title(title)
    plt.ticklabel_format(style="plain")
    plt.show()
```

## Проверка гипотез (#hypothesis_testing)
<a id="data"></a>

Продакт-менеджеры сервиса хотят увеличить количество пользователей с подпиской. Для этого они будут проводить различные акции, но сначала нужно выяснить несколько важных моментов.  

Первая гипотеза, котрую необходимо проверить — пользователи с подпиской тратят больше времени на поездки и могут быть «выгоднее» для компании. Для её проверки исследую данные о продолжительности каждой сессии — отдельно для подписчиков и тех, у кого нет подписки.  

Для проверки гипотезы сравню среднюю продолжительность использования приложения для пользователей с подпиской и без неё.  
H0: средняя продолжительность использования приложения не увеличивается при покупке подписки.  
H1: средняя продолжительность использования приложения увеличивается при покупке подписки.

<div class="alert alert-block alert-success">✔️
    

__Комментарий от ревьюера №1__

Верная интерпретация нулевой и альтернативной гипотез
</div>

```python
print(
    f' Средняя продолжительность использования приложения пользователей с подпиской: {np.mean(users_subscribes["duration"])}'
)
print(
    f' Средняя продолжительность использования приложения пользователей без подписки: {np.mean(users_unsubscribes["duration"])}'
)

# уровень статистической значимости
alpha = 0.05

results = st.ttest_ind(
    users_subscribes["duration"], users_unsubscribes["duration"], alternative="greater"
)
# выведем полученное p-value
print("p-значение:", results.pvalue)
```

**Среднее значение — на 1,15 минут (18,54 - 17,39) или на 6,2%.**  
**При статистической значимости 5% такое изменение действительно существенно, а значит нулевая гипотеза отвергается.**  

Вторая гипотеза, котрую необходимо проверить — среднее расстояние, которое проезжают пользователи с подпиской за одну поездку, не превышает 3130 метров.  

Для проверки гипотезы необходимо сравнить среднее расстояние пользователей с подпиской с предположенным в нулевой гипотезе числом (t-тест).  
H0: среднее расстояние, которое проезжают пользователи с подпиской за одну поездку превышает 3130 метров.  
H1: среднее расстояние, которое проезжают пользователи не больше 3130 метров.

```python
print(
    f'Среднее расстояние, которое проезжают пользователи с подпиской за одну поездку: {np.mean(users_subscribes["distance"])}'
)

screens = users_subscribes["distance"]

prev_screens_value = 3130

alpha = 0.05  # уровень статистической значимости

results = st.ttest_1samp(screens, prev_screens_value, alternative="greater")

print("p-значение:", results.pvalue)

if results.pvalue > alpha:
    print("Отвергаем нулевую гипотезу")
else:
    print("Не отвергаем нулевую гипотезу")
results
```

<div class="alert alert-block alert-warning">📝
    

__Комментарий от ревьюера №1__

Смотри, поскольку оптимальное расстояние по износу составляет 3130 м, значит большее расстояние приводит к повышенному износу самоката. Следовательно имеет смысл проверить одностороннюю гипотезу, что расстояние, которое проезжают пользователи с подпиской не превышает оптимального расстояния в 3130 м.

То есть в статистическом тесте стоит поставить параметр alternative = 'greater' и изменить Н1 на знак `больше`

Третья гипотеза, котрую необходимо проверить — помесячная выручка от пользователей с подпиской по месяцам выше, чем выручка от пользователей без подписки.  

Для проверки гипотезы необходимо сравнить две независимые выборки о выручке пользователей с подпиской и без.  
H0: Средняя помесячная выручка от группы пользователей с подпиской равна средней выручке пользователей без подписки.  
H1: Средняя помесячная выручка от группы пользователей с подпиской выше, чем выручка от пользователей без подписки.

```python
# получаем датафреймы с выручкой  пользователей с подпиской и без
rides_subscibes = rides_df[rides_df["subscription_type"] == "ultra"]
rides_unsubscibes = rides_df[rides_df["subscription_type"] == "free"]

# вывожу средние помесячные выручки для каждой из групп пользователей
print(rides_subscibes["revenue"].mean())
print(rides_unsubscibes["revenue"].mean())

alpha = 0.05  # уровень статистической значимости
# если p-value окажется меньше него, отвергнем гипотезу

results = st.ttest_ind(
    rides_subscibes["revenue"], rides_unsubscibes["revenue"], alternative="greater"
)

print("p-значение:", results.pvalue)

if results.pvalue < alpha:
    print("Отвергаем нулевую гипотезу")
else:
    print("Не получилось отвергнуть нулевую гипотезу")
```
