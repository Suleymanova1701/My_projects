## Описание проекта (#description_project)
<a id="description_project"></a>

Сеть фитнес-центров «Культурист-датасаентист» разрабатывает стратегию взаимодействия с пользователями на основе аналитических данных.  

Распространённая проблема фитнес-клубов и других сервисов — отток клиентов. Как понять, что пользователь больше не с вами? Можно записать в отток тех, кто попросил закрыть договор или удалил аккаунт. Однако клиенты не всегда уходят демонстративно: чаще перестают пользоваться сервисом тихо.  

Индикаторы оттока зависят от специфики отрасли. Когда пользователь редко, но стабильно закупается в интернет-магазине — не похоже, что он «отвалился». А вот если две недели не заходит на канал с ежедневно обновляемым контентом, дела плохи: подписчик заскучал и, кажется, оставил вас.  

Для фитнес-центра можно считать, что клиент попал в отток, если за последний месяц ни разу не посетил спортзал. Конечно, не исключено, что он уехал на Бали и по приезде обязательно продолжит ходить на фитнес. Однако чаще бывает наоборот. Если клиент начал новую жизнь с понедельника, немного походил в спортзал, а потом пропал — скорее всего, он не вернётся.  

Чтобы бороться с оттоком, отдел по работе с клиентами «Культуриста-датасаентиста» перевёл в электронный вид множество анкет пользователей. Задача — провести анализ и подготовить план действий по удержанию клиентов.

    1) выделить целевые группы клиентов;

Для каждого полученного кластера посчитайте долю оттока (методом groupby()). Отличаются ли они по доле оттока? Какие кластеры склонны к оттоку, а какие — надёжны?
5. Сформулировать основные выводы и предложить рекомендации для стратегии взаимодействия с пользователями и их удержания.

#### Функции для предобработки данных

````python
def get_count_and_share(df: pd.DataFrame, groupby: str = "churn") -> pd.DataFrame:
    """

    Функция формирует и выводит сводную таблицу по одному из атрибутов (город по умолчанию),

    в которой отражает количество и долю строк каждого из этих атрибутов.
    """

    # считаю количество пользователей

    count_and_share = df.groupby(groupby).agg({groupby: "count"})

    # переименовываю столбец

    count_and_share.columns = ["count"]

    # считаю долю

    count_and_share["%_share_of_"] = round(
        (count_and_share["count"]).div(df["churn"].count(), axis=0), 4
    )

    count_and_share = count_and_share[count_and_share["%_share_of_"] >= 0.001]

    # переименовываю столбцы финального датафрейма

    count_and_share.columns = ["count_of_" + groupby, "%_share_of_" + groupby]

    # сортирую финальный датафрейм по убыванию

    count_and_share = count_and_share.sort_values(
        by="%_share_of_" + groupby, ascending=False
    )

    count_and_share = count_and_share.style.format(
        {"%_share_of_" + groupby: "{:.2%}".format}
    ).background_gradient(cmap="Greens")
    return count_and_share
````

## Модель прогнозирования оттока клиентов. (#customer_churn_prediction_model)
<a id="data"></a>

В этом разделе построю модель бинарной классификации клиентов, где целевой признак — факт оттока клиента в следующем месяце. Для этого:  
- разобью данные на обучающую и валидационную выборку функцией `train_test_split()`;  
- обучу модель на train-выборке двумя способами:
    - логистической регрессией;
    - случайным лесом;  
- оценю метрики `accuracy`, `precision` и `recall` для обеих моделей на валидационной выборке;  
- сделаю выводы по сравнению моделей.

````python
# разбиваю выборки на обучающую и валидационную
X = gym_churn.drop(["churn"], axis=1)
y = gym_churn["churn"]


# разбиваю данные на обучающую и валидационную в пропорции 80/20
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

scaler = StandardScaler()
X_train_st = scaler.fit_transform(X_train)
X_test_st = scaler.transform(X_test)

lr_model = LogisticRegression(random_state=0)
lr_model.fit(X_train, y_train)
lr_predictions = lr_model.predict(X_test)
lr_probabilities = lr_model.predict_proba(X_test)[:, 1]
print_all_metrics(
    y_test,
    lr_predictions,
    lr_probabilities,
    title="Метрики для модели логистической регрессии:",
)

rf_model = RandomForestClassifier(n_estimators=100, random_state=0)
rf_model.fit(X_train, y_train)
rf_predictions = rf_model.predict(X_test)
rf_probabilities = rf_model.predict_proba(X_test)[:, 1]
print_all_metrics(
    y_test,
    rf_predictions,
    rf_probabilities,
    title="Метрики для модели случайного леса:",
)

log_reg = pd.DataFrame({"feature": X_train.columns, "coef": lr_model.coef_[0]})
log_reg = log_reg.sort_values(by="coef", ascending=True)
print("\nКоэффициенты признаков в оптимальной функции логистической регрессии:")
print(log_reg)
````

1. Мы делим наши данные на три выборки, обучающую, валидационну и тестовую (обычно в соотношении 3:1:1)
2. Циклом проходим по нашим параметрам, для каждого набора параметров обучаем модель на обучающей выборке и смотрим на метрики на валидационной
3. Выбираем те гиперпараметры, при которых выбранная метрика на валидационной выборке будет максимальной
4. Объединяем обучающую и валидационну выборки в новую, обучающую выборку, назовем ее обучающая_2
5. Обучаем модель с подобранными гиперпараметрами на выборке обучающая_2
6. Смотрим на метрики на тестовой выборке
