## Описание проекта (#description_project)
<a id="description_project"></a>

Интернет-магазин товаров для дома «Пока всё ещё тут» в срочном порядке ищет аналитиков для сегментации покупателей и подготовки базы для рассылок.

*Задача исследования:*
1. Сегментировать покупателей по профилю потребления для осуществления рекламных рассылок.
2. Выяснить в каком сегменте покупают чаще, больше и на бОльшую сумму и какой сегмент наименее привлекателен с точки зрения продаж.
3. Выяснить клиентам какого сегмента, когда и какие категории товара рассылать в рекламных предложениях (товар в датасете по категориям не разбит)
4. Выяснить как и для какого кластера можно попытаться увеличить количество продаваемого товара, повысить возвращаемость покупателей и т.д.


*Цель исследования:*
1. Подробный анализ продаж и кластеризация товаров.
2. Выявление наиболее популярных и выгодных кластеров товаров.
3. Создание аватаров пользователей, описание каждого из аватаров.
4. Выявление наиболее привлекательных сценариев продаж для каждого из сегментов пользователй.
5. Подготовка сценария увеличения объёма продаж для каждого из сегментов ЦА.
6. Подготовка сценария привлечения и удержания клиента каждого из сегментов ЦА.

*Ход исследования:*
1. Настройка параметров анализа.
2. Загрузка и оценка качества данных.
3. Предобработка данных: 
    - проверка типов данных;
    - корректировка наименований столбцов таблиц при необходимости;
    - дополнение таблицы дополнительными расчётами;
    - обработка пропусков и дубликатов с выводами;
    - формулировка выводов по полученному набору данных.
4. Исследовательский анализ данных: 
- визуализация продаж;
- кластеризация продаж по:
    - средней цене товара;
    - среднему количеству купленных товаров;
    - среднему чеку продажи.
5. Портрет пользователя:
    В этом разделе определю для пользователя каждого сегмента:  
    - ТОП-5 популярных категорий товаров для каждого сегмента покупателей;  
    - частоту покупок каждого из сегментов покупателей;  
    - средний чек продажи для каждого из сегментов
    - средний чек одной покупки;
    - средний чек сегмента;
    - средняя частота покупок;
    - категории популярных товаров;  
    - другие особенности потребительского поведения.  
    Сделаю выводы.
6. Проверка гипотезы:
    - о равенстве средних чеков сегментов покупателей:  
        H0: средние чеки сегмента 1 и сегмента 2 равны;  
        H1: средние чеки сегментов 1 и 2 отличаются.  
    - о сумме выручки:  
        Н0: средняя выручка потребителей сегмента 1 больше средней выручки потребителей сегмента 2;  
        Н1: средняя выручка потребителей сегмента 1 не больше средней выручки потребителей сегмента 2.
7. Формулирование общего вывода.

### Функции для анализа данных

````python
def get_count_and_share(
        df: pd.DataFrame, 
        title: str, 
        xlabel: str, 
        ylabel:str, 
        groupby: str="month", 
        aggfuncby: str="order_id", 
        aggfunc: str='count'
        ) -> pd.DataFrame:
    
    """
    Функция формирует и выводит сводную таблицу по одному из атрибутов (месяц по умолчанию),
    в которой отражает количество и долю строк каждого из этих атрибутов.
    """

    # создаю сводную таблицу по выбранным параметрам
    count_and_share = df.pivot_table(
        index=groupby, 
        values=aggfuncby, 
        aggfunc=aggfunc
        )

    # переименовываю столбец
    count_and_share.columns = [aggfunc]

    #строю график динамики количества
    count_and_share.plot()
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.title(title)
    plt.show()
    
    # считаю долю
    if aggfunc=='count':
        count_and_share["%_share_of_"] = round(
            (
                count_and_share[aggfunc]).div(df[aggfuncby].count(), axis=0
                                              ), 4)
    elif aggfunc=='sum':
        count_and_share[aggfunc] = round(
                count_and_share[aggfunc], 2)
        count_and_share["%_share_of_"] = round(
            (
                count_and_share[aggfunc]).div(df[aggfuncby].sum(), axis=0
                                              ), 4)
   
    # переименовываю столбцы финального датафрейма
    count_and_share.columns = [aggfunc + "_of_" + aggfuncby, "%_share_of_" + aggfuncby]

    # сортирую финальный датафрейм по убыванию
    count_and_share = count_and_share.sort_values(
        by="%_share_of_" + aggfuncby, ascending=False
    )
    

    count_and_share = count_and_share.style.format(
        {"%_share_of_" + aggfuncby: "{:.2%}".format}
    ).background_gradient(cmap="YlGnBu")
    
    return count_and_share
````

### Функции для проверки гипотез

````python
def shapiro_test(data: pd.DataFrame, attribute: str, alpha: float=0.05) -> str:
    '''
    Функция проверяет распределения значений принимаемого датасета на нормальность критерием Шапиро-Уилка
    # Нулевая гипотеза: значения проверяемого датафрейма каждого из кластеров распределены нормально
    # Альтернативная гипотеза: значения проверяемого датафрейма распределены не нормально
    '''
    for i in range(3):
        stat, p_value = st.shapiro(data[data['cluster'] == i][attribute])
        print(f'p-значение cluster {i}: {p_value}')  # Значение атрибута покупателей сегмента {i}: 
        if p_value < alpha:
            print(f'Отвергаем нулевую гипотезу: значения распределены не нормально' )
        else:
            print('Не получилось отвергнуть нулевую гипотезу, значения распределены нормально') 
````

````python
def get_hypothesis_testing_results(attribute: str) -> str:
    '''
    Функция проверяет тестирует гипотезы методом Манна-Уитни
    '''


    # создаю переменные по каждому сегменту с нужным атрибутом
    segment_attribute = []
    for i in range(3):
        segment_attribute.append(df[df['cluster'] == i][attribute])

    # вывожу средний чек по каждому сегменту
    for i in range(3):
        if attribute == 'avg_revenue':
            print(
                f'Средний чек покупателей сегмента {i}:  {np.mean(segment_attribute[i])}\n',
            )
        elif attribute == 'avg_number_of_products':
            print(
                f'Среднее число покупок покупателей сегмента {i}:  {np.mean(segment_attribute[i])}\n',
            )
        else:
            print(
                f'Среднее значение признака {attribute} покупателей сегмента {i}:  {np.mean(segment_attribute[i])}\n',
            )
    
    # уровень статистической значимости
    m = 3 # число сегментов
    alpha = 0.05 # исходный уровень значимости

    # считаю поправку
    alpha_sidak = 1 - (1 - alpha) ** (1 / m) 
    alpha_sidak
    print(
        f'Уровень статистической значимости с поправкой на количество тестов: {str(alpha_sidak)}\n'
        )

    for s1, s2 in itertools.combinations(range(3), 2):
        results = st.mannwhitneyu(
            segment_attribute[s1], segment_attribute[s2]
        )
        # выведем полученное p-value
        print(f"p-значение segment_{s1}, segment_{s2}:", results.pvalue)
        if results.pvalue < alpha_sidak:
            print("Отвергаем нулевую гипотезу")
        else:
            print("Не отвергаем нулевую гипотезу")
        print(results, '\n')
````

Добавлю параметр частоты покупок для каждого из клиентов, сгруппировав их по id покупателя и дате покупки. 

````python
# создаю отдельный датафрейм с активными покупателями
active_customers = (
    df.groupby('customer_id')
    .filter(lambda x: x['order_id']
            .nunique() > 1).sort_values(by=['customer_id', 'date'])
)

# Расчет времени между покупками для каждого клиента
active_customers['previous_date'] = active_customers.groupby('customer_id')['date'].shift(1)
active_customers['days_between_purchases'] = (active_customers['date'] - active_customers['previous_date']).dt.days

# Расчет среднего времени между покупками для каждого клиента
average_days_between_purchases = active_customers.groupby('customer_id')['days_between_purchases'].mean()
average_days_between_purchases.columns = ['customer_id', 'frequency_of_purchases'] 

print(f"Средняя частота покупок активных клиентов: {average_days_between_purchases.mean():.2f} дней")
print(f"Медианная частота покупок активных клиентов: {average_days_between_purchases.median():.2f} дней")
# print(f"Мода частоты покупок активных клиентов: {average_days_between_purchases.value_counts().head(1):.2f} дней")


# Визуализация распределения времени между покупками
plt.figure(figsize=(10, 6))
plt.hist(average_days_between_purchases, bins=30, color='skyblue', edgecolor='black')
plt.title('Распределение частоты покупок активных клиентов')
plt.xlabel('Частота покупок в днях')
plt.ylabel('Количество клиентов')
plt.grid(axis='y')

plt.show()
average_days_between_purchases
````

Гистограмма расределения средней частоты покупок показывает что большинство клиентов сделали 1 покупку, однако достаочно большое количество покупателей со средней частотой покупки более 100 дней.

````python
# создадим пустой список для записи показателя WCSS (ошибки)
wcss = []
 
# воспользуемся новой для нас функцией range(), она возвращает последовательность чисел
# в данном случае числа от 1 до 10
for i in range(1, 11):
    
    # настроим параметры модели (подробнее об этом расскажу чуть дальше)
    kmeans = KMeans(n_clusters = i, init = 'k-means++', max_iter = 300, n_init = 10, random_state = 42)
 
    # обучим модель на данных с разным количеством кластеров
    kmeans.fit(X_sc)
 
    # для каждого кластера рассчитаем ошибку (атрибут inertia_) и поместим в список
    wcss.append(kmeans.inertia_)
 
# зададим размер график
plt.figure(figsize = (10,6))
# передадим функции plot() последовательность кластеров и ошибку WCSS
plt.plot(range(1, 11), wcss)
# и добавим подписи
plt.title('Выбор количества кластеров методом локтя')
plt.xlabel('Количество кластеров')
plt.ylabel('WCSS')
plt.show()
````

## Проверка гипотез (#hypothesis_testing)
<a id="description_project"></a>

В этом разделе проверю гипотезы:

- о наличии статистически значимой разницы средних чеков сегментов покупателей:  
        H0: нет статистически значимой разницы средних чеков между сегментами;

- о количестве покупок:  
        Н0: нет статистически значимой разницы среднего количества покупок сегментов;  
        Н1: между средним количеством покупок сегментов есть статистически значимая разница.
